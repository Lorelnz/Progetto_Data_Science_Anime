1) Esiste una differenza statisticamente significativa nelle valutazioni in base al formato di rilascio (TV, Movie, OVA, ecc.)?

```python
import pandas as pd
import numpy as np
import scipy.stats as sp
import statsmodels.api as sm
import matplotlib.pyplot as plt
import seaborn as sns

df=pd.read_pickle('10KAnime.pickle')
```

```python
print(df['score'].min())

#inizializzo lo stile
sns.set_theme(style='whitegrid', context='talk')

plt.figure(figsize=(8.5, 5))

#grafico
sns.violinplot(data=df, x='score', y='type', hue='type', 
               palette='Set2', inner='quartile', linewidth=1.5, legend=False)

#titoli degli assi
plt.title('Distribuzione della valutazione in base al formato\n(Analisi della varianza - ANOVA)', weight='bold')
plt.xlabel('Voto medio (Voto: 5 - 10)', weight='bold')
plt.ylabel('Formato editoriale', weight='bold')
#ottimizzo gli spazi
plt.xlim(5, 10)

#salvataggio
plt.savefig("Figures/1.png", bbox_inches='tight')
plt.savefig("Figures/1.svg", bbox_inches='tight')

plt.tight_layout()
plt.show()
```
